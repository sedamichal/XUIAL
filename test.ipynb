```python
import cv2
from matplotlib import pyplot as plt
from src.pso_processors import PSOKapurEntropyProcessor, PSOBlockClusteringProcessor

def show_image(img, caption):    
    i = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.imshow(i)
    plt.title(caption)
    plt.show()

def process_single_kapur(pop, iters, img):
    processor = PSOKapurEntropyProcessor(w=0.5, c1=1.5, c2=1.5, pop=pop, iters=iters)
    mask = processor.get_mask(img)
    show_image(mask, f"{pop}x{iters}" )

def process_single_block(pop, iters, img):
    processor = PSOBlockClusteringProcessor(w=0.5, c1=1.5, c2=1.5, pop=pop, iters=iters, block_size=8)
    mask = processor.get_mask(img)
    show_image(mask, f"{pop}x{iters}" )
```

```python
import cv2
from src.pso_processors import PSOKapurEntropyProcessor, PSOBlockClusteringProcessor

image = cv2.imread("7.jpg", cv2.IMREAD_GRAYSCALE)

I_MIN = 5
P_MIN = 10
I_MAX = 6
P_MAX = 11
```

```python
for pop in range(P_MIN, P_MAX):
    for iters in range(I_MIN, I_MAX):
        process_single_kapur(pop, iters, image)
```

```python
for pop in range(P_MIN, P_MAX):
    for iters in range(I_MIN, I_MAX):
        process_single_block(pop, iters, image)
```
